# Representation Learning and Generative Learning Using Autoencoders and GANs

Autoencoders and GANs are both unsupervised, they both learn dense represenations, they can both be used as generative models, and they have many similar applications. However, they work veyr differently:
* Autoencoders simply learn to copy their inputs to their outputs. This may sound like a trivial task, but we will see that constraining the network in various ways can make it rather difficult. For example, you can limit the size of the latent representations, or you can add noise to the inputs and train the network to recover the original inputs. These constraints prevent the autoencoder from trivially copying the inputs directly to the outputs, which forces it to learn efficient ways of representing the data. In short, the codings are byproducts of the autoencoder learning the idenity function under some constraints.
* GANs are composed of two neural networks: a *generator* that tries to generate data that looks similar to the training data, and a *discriminator* that tries to tell real data from fake data. This architecture is very original in Deep Learning in that the generator and the discriminator compete against each otherduring training: the generator is often compared to a criminal trying to make realistic counterfeit money, while the discriminator is like the police investigator trying to tell real money from fake. *Adversarial training* (training competing neural networks) is widely considered as one of the most important ideas in recent years. In 2016, Yann LeCun even said that it was "the most interesting idea in the last decade of Machine Learning"

In this chapter we will start by exploring in more depth how autoencoders work and how to **use them for dimensionality reduction, feature extraction, unsupervised pretraining, or as generative models**. This will naturally lead us to GANs

## Efficient Data Representations

Famously studied by William Chase and Herbert Simon in the early 1970s, chess players were able to memorize the positions of all the pieces in a game by looking at the board for just five seconds, a task that most people would find impossible. However, this was only the case when the pieces were placed in realistic positions, not when the pieces were placed randomly. Chess experts don't have a much better memory than you or I; they just see chess patterns more easily, thanks to their experience with the game. Noticing patterns helps them store information efficiently.

Just like the chess players in this memory experiment, an autoencoder looks at the inputs, converts them to an efficient latent representation, and then spits out something hopefully looks very close to the inputs. an autoencoder is always composed of two parts: an *encoder* (or *recognition network*) that converts the inputs to a latent representation, followed by a *decoder* (or *generative network*) that converts the internal representation to the outputs. An autoencoder typically has the same architecture as a Multi-Layer Perceptron (MLP), except that the number of neurons in the output layer must be equal to the number of inputs.

The outputs are often called the *reconstructions* because the autoencoder tries to reconstruct the inputs, and the cost function contains a *reconstruction loss* that penalizes the model when the reconstructions are different from the inputs. Because the internal representation has a lower dimensionality than the input data (2D instead of 3D), the autoencoder is said to be *undercomplete*. An undercomplete autoencoder cannot trivially copy the its inputs to the codings, yet it must find a way to output a copy of its inputs. It is forced to learn the most important features in the input data.

## Performing PCA with an Undercomplete Linear Autoencoder

If the autoencoder uses only linear activations and the cost function is the mean squared error then it ends up performing Principal Component Analysis. The following code builds a simple linear autoencoder to perform PCA on a 3D dataset, projecting it to 2D. There are a few things to note:

* We organized the autoencoder into two subcomponents: the encoder and the decoder. Both are regular Sequential models with a single Dense layer each, and the autoencoder is a Sequential model containing the encoder followed by the decoder (remember that a model can be used as a layer in another model).
* The autoencoder's number of outputs is equal to the number of inputs.
* To perform simple PCA, we do not use any activation function and the cost function is MSE. We will see more complex autoencoders soon.

You can think of autoencoders as a form of self-supervised learning (i.e. using a supervised learning technqiue with automatically generated labels, in this case simply equal to the inputs).

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.optimizers import SGD
import numpy as np
from sklearn.model_selection import train_test_split

np.random.seed(42)

n_samples = 1000

# Generate 2D latent structure
z1 = np.random.uniform(-1, 1, n_samples)
z2 = np.random.uniform(-1, 1, n_samples)

# Embed into 3D plane with noise
X = np.column_stack([
    z1,
    z2,
    0.5 * z1 + 0.25 * z2
])

X += 0.05 * np.random.randn(n_samples, 3)

# Split: 70% train, 15% val, 15% test
X_train, X_temp = train_test_split(X, test_size=0.30, random_state=42)
X_val, X_test = train_test_split(X_temp, test_size=0.50, random_state=42)

encoder = Sequential([
    Input(shape=(3,)),
    Dense(2)
])

decoder = Sequential([
    Input(shape=(2,)),
    Dense(3)
])

autoencoder = Sequential([
    encoder,
    decoder
])

autoencoder.compile(
    loss='mse',
    optimizer=SGD(learning_rate=0.1)
)

history = autoencoder.fit(X_train, X_train, epochs=20)
codings = encoder.predict(X_train)
```

```python
import matplotlib.pyplot as plt

codings = encoder.predict(X_train)

fig = plt.figure(figsize=(12, 5))

# Original 3D data
ax1 = fig.add_subplot(1, 2, 1, projection="3d")
ax1.scatter(
    X_train[:, 0],
    X_train[:, 1],
    X_train[:, 2],
    s=10,
    alpha=0.6
)
ax1.set_title("Original 3D Data")
ax1.set_xlabel("x1")
ax1.set_ylabel("x2")
ax1.set_zlabel("x3")

# 2D encoded representation
ax2 = fig.add_subplot(1, 2, 2)
ax2.scatter(
    codings[:, 0],
    codings[:, 1],
    s=10,
    alpha=0.6
)
ax2.set_title("Autoencoder PCA-Style 2D Projection")
ax2.set_xlabel("coding 1")
ax2.set_ylabel("coding 2")

plt.tight_layout()
plt.show()
```

## Stacked Autoencoders

Just like other neural networks we have discussed, autoencoders can have multiple hidden layers. In this case they are called *stacked autoencoders* (or *deep autoencoders*). Adding more layers helps the autoencoder learn more complex codings. That said, one must be careful not to make the autoencoder too powerful. With too many available parameters the autoencoder will essentially create a 1-1 map of the training data and fail to learn any useful data representations.

The architecture of a stacked autoencoder is typically symmetrical with regard to the central hidden layer. To put it simply, it looks like a sandwhich. For example, an autoencoder for MNIST may have 784 inputs, followed by a hidden layer with 100 neurons, then a central hidden layer of 30 neurons, then another hidden layer with 100 neurons, and an output layer with 784 neurons.

### Implementing a Stacked Autoencoder Using Keras

The following code builds a stacked autoencoder for Fashion MNIST. Let's go through the code:

* Just like earlier, we split the autoenoder model into two submodels: the encoder and the decoder
* The encoder takes 28x28-pixel grayscale images, flattens them so that each image is represented as a vector of size 784, then processes these vectors through two Dense layers of diminishing sizes (100 units then 30 units), both using the SELU activation function (you may want to add LeCun normal initialization as well, but the network is not very deep so it won't make a big difference). For each input image, the encoder outputs a vector of size 30
* The decoder takes codings of size 30 (output by the encoder) and processes them through two Dense layers of increasing sizes (100 units then 784 units), and it reshapes the final vectors into 28x28 arrays so the decoder's outputs have the same shape as the encoder's inputs.
* When compiling the stacked autoencoder, we use the binary cross-entropy loss instead of the mean squared error. We are treating the reconstruction task as a multilabel binary classification problem: each pixel intensity represents the probability that the pixel should be black. Framing it this way (rather than as a regression problem) tends to make the model converge faster. You might be tempted to use the accuracy metric, but it would not work properly, since this metric expects the labels to be either 0 or 1 for each pixel. You can easily work around this problem by creating a custom metric that commputes the accuracy after rounding the targets and predictions to 0 or 1.

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Flatten, Dense, Reshape
from tensorflow.keras.initializers import LecunNormal
from tensorflow.keras.optimizers import Nadam


# -----------------------------
# 1. Load and prepare Fashion MNIST
# -----------------------------

(X_train_full, y_train_full), (X_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

# Scale pixel intensities to [0, 1]
X_train_full = X_train_full.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Validation split
X_train = X_train_full[:-5000]
X_val = X_train_full[-5000:]
y_val = y_train_full[-5000:]


# -----------------------------
# 2. Custom rounded accuracy metric
# -----------------------------

class RoundedAccuracy(tf.keras.metrics.Metric):
    def __init__(self, name="rounded_accuracy", **kwargs):
        super().__init__(name=name, **kwargs)
        self.binary_accuracy = tf.keras.metrics.BinaryAccuracy()

    def update_state(self, y_true, y_pred, sample_weight=None):
        y_true_rounded = tf.round(y_true)
        y_pred_rounded = tf.round(y_pred)

        self.binary_accuracy.update_state(
            y_true_rounded,
            y_pred_rounded,
            sample_weight=sample_weight
        )

    def result(self):
        return self.binary_accuracy.result()

    def reset_state(self):
        self.binary_accuracy.reset_state()


# -----------------------------
# 3. Build encoder and decoder
# -----------------------------

encoder = Sequential([
    Input(shape=(28, 28)),
    Flatten(),
    Dense(100, activation="selu", kernel_initializer=LecunNormal()),
    Dense(30, activation="selu", kernel_initializer=LecunNormal())
], name="encoder")

decoder = Sequential([
    Input(shape=(30,)),
    Dense(100, activation="selu", kernel_initializer=LecunNormal()),
    Dense(28 * 28, activation="sigmoid"),
    Reshape((28, 28))
], name="decoder")


# -----------------------------
# 4. Build stacked autoencoder
# -----------------------------

stacked_autoencoder = Sequential([
    encoder,
    decoder
], name="stacked_autoencoder")


# -----------------------------
# 5. Compile model
# -----------------------------

stacked_autoencoder.compile(
    loss="binary_crossentropy",
    optimizer=Nadam(),
    metrics=[RoundedAccuracy()]
)


# -----------------------------
# 6. Train model
# -----------------------------

history = stacked_autoencoder.fit(
    X_train,
    X_train,
    epochs=10,
    validation_data=(X_val, X_val)
)


# -----------------------------
# 7. Reconstruct images
# -----------------------------

X_reconstructions = stacked_autoencoder.predict(X_test[:5])
codings = encoder.predict(X_test[:5])
```

### Visualizing the Reconstructions

One way to ensure that an autoencoder is properly trained is to compare the input and the outputs: the differences should not be too significant. Let's plot a few images from the validation set, as well as their reconstructions. What we notice is the reconstructions are recognizable but too noisy. We may need to train the model longer or make the encoder and decoder deeper, or make the codings larger. But if we make the network too powerful, it will manage to make perfect reconstructions without having learned any useful patterns in the data. For now, let's go with this model.

```python
import matplotlib.pyplot as plt

def plot_reconstructions(model, data, starting_slice_index=0):
    """
    Plots 5 original images in row 1 and their reconstructions in row 2.

    Parameters
    ----------
    model : tf.keras.Model
        Trained autoencoder model.
    data : np.ndarray or tf.Tensor
        Image data shaped like (n_samples, 28, 28).
    starting_slice_index : int
        Starting index for selecting 5 images.
    """

    images = data[starting_slice_index : starting_slice_index + 5]
    reconstructions = model.predict(images, verbose=0)

    fig, axes = plt.subplots(2, 5, figsize=(10, 4))

    for i in range(5):
        # Original images
        axes[0, i].imshow(images[i], cmap="binary")
        axes[0, i].axis("off")
        axes[0, i].set_title(f"Original {starting_slice_index + i}")

        # Reconstructed images
        axes[1, i].imshow(reconstructions[i], cmap="binary")
        axes[1, i].axis("off")
        axes[1, i].set_title("Reconstructed")

    plt.tight_layout()
    plt.show()

plot_reconstructions(
    model=stacked_autoencoder,
    data=X_val,
    starting_slice_index=0
)
```

### Unsupervised Pretraining Using Stacked Autoencoders

As we discussed in Chapter 11, if you are tackling a complex supervised task but you do not have a lot of labeled training data, one solution is to find a neural network that performs a similar task and reuse its lower layers. This makes it possible to train a high-performance model using little training data because your neural network won't have to learn all the low-level features; it will just reuse the feature detectors learned by the existing network. 

Similarly, **if you have a large dataset but most of it is unlabeled, you can first train a stacked autoencoder using all the data, then reuse the lower layers to create a neural newtork for your actual task and train it using the labeled data. When training the model, if you really don't have much labeled training data, you may want to freeze the pretrained layers (at least the lower ones).**

Having plenty of unlabeled data and little labeled data is common. Building a large unlabeled dataset is cheap (e.g. a simple script can download millions of images off the internet), but labeling those images an usually be done reliably only by humans. Labeling instances is time-consuming and costly, so it's normal to have only a few thousands human-labeled instances.

**There is nothing special about the implementation: just train an autoencoder using all the training data then reuse its encoder layers to create a new neural network.**

### Tying Weights

When an autoencoder is neatly symmetrical like the one we just built a common technique is to *tie* the weights of the decoder layers to the weights of the encoder layers. This halves the number of weights in the model speeding up training and limiting the risk of overfitting.

To tie weights between layers define a customer layer like the one shown below. This custom layer acts like a regular Dense layer but it uses another Dense layer's weights, transposed (setting transpose_b=True is equivalent to transposing the second argument, but it's more efficient as it performs the transposition on the fly within the matmul() operation). However, it uses its own bias vector. After, we can build a new stacked autoencoder, much like the previous one, but with the decoder's Desne layers tied to the encoder's Dense layers

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Flatten, Dense, Reshape, Layer
from tensorflow.keras.initializers import LecunNormal
from tensorflow.keras.optimizers import Nadam


# -----------------------------
# 1. Load and prepare Fashion MNIST
# -----------------------------

(X_train_full, y_train_full), (X_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

X_train_full = X_train_full.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

X_train = X_train_full[:-5000]
X_val = X_train_full[-5000:]

y_train = y_train_full[:-5000]
y_val = y_train_full[-5000:]


# -----------------------------
# 2. Custom rounded accuracy metric
# -----------------------------

class RoundedAccuracy(tf.keras.metrics.Metric):
    def __init__(self, name="rounded_accuracy", **kwargs):
        super().__init__(name=name, **kwargs)
        self.binary_accuracy = tf.keras.metrics.BinaryAccuracy()

    def update_state(self, y_true, y_pred, sample_weight=None):
        y_true_rounded = tf.round(y_true)
        y_pred_rounded = tf.round(y_pred)

        self.binary_accuracy.update_state(
            y_true_rounded,
            y_pred_rounded,
            sample_weight=sample_weight
        )

    def result(self):
        return self.binary_accuracy.result()

    def reset_state(self):
        self.binary_accuracy.reset_state()


# -----------------------------
# 3. Custom tied Dense layer
# -----------------------------

class DenseTranspose(Layer):
    """
    Dense-like layer whose kernel is tied to another Dense layer's kernel,
    using the transpose of that kernel.

    If tied_to has kernel shape:
        (input_dim, units)

    then this layer uses:
        tied_to.kernel.T

    so inputs must have shape:
        (..., units)

    and outputs will have shape:
        (..., input_dim)
    """

    def __init__(self, tied_to, activation=None, **kwargs):
        super().__init__(**kwargs)
        self.tied_to = tied_to
        self.activation = tf.keras.activations.get(activation)

    def build(self, input_shape):
        output_dim = self.tied_to.kernel.shape[0]

        self.biases = self.add_weight(
            name="biases",
            shape=(output_dim,),
            initializer="zeros",
            trainable=True
        )

        super().build(input_shape)

    def call(self, inputs):
        z = tf.matmul(inputs, self.tied_to.kernel, transpose_b=True)
        z = z + self.biases

        if self.activation is not None:
            z = self.activation(z)

        return z

    def get_config(self):
        config = super().get_config()
        config.update({
            "activation": tf.keras.activations.serialize(self.activation),
        })
        return config


# -----------------------------
# 4. Build tied-weights autoencoder
# -----------------------------

# Encoder Dense layers must be named/referenced so decoder can tie to them
dense_1 = Dense(
    100,
    activation="selu",
    kernel_initializer=LecunNormal(),
    name="encoder_dense_100"
)

dense_2 = Dense(
    30,
    activation="selu",
    kernel_initializer=LecunNormal(),
    name="encoder_dense_30"
)

encoder = Sequential([
    Input(shape=(28, 28)),
    Flatten(),
    dense_1,
    dense_2
], name="encoder")

decoder = Sequential([
    Input(shape=(30,)),
    DenseTranspose(dense_2, activation="selu", name="decoder_tied_dense_100"),
    DenseTranspose(dense_1, activation="sigmoid", name="decoder_tied_dense_784"),
    Reshape((28, 28))
], name="decoder")

tied_stacked_autoencoder = Sequential([
    encoder,
    decoder
], name="tied_stacked_autoencoder")


# -----------------------------
# 5. Compile
# -----------------------------

tied_stacked_autoencoder.compile(
    loss="binary_crossentropy",
    optimizer=Nadam(),
    metrics=[RoundedAccuracy()]
)


# -----------------------------
# 6. Train
# -----------------------------

history = tied_stacked_autoencoder.fit(
    X_train,
    X_train,
    epochs=10,
    validation_data=(X_val, X_val)
)


# -----------------------------
# 7. Use encoder / reconstruct images
# -----------------------------

codings = encoder.predict(X_test[:5], verbose=0)
reconstructions = tied_stacked_autoencoder.predict(X_test[:5], verbose=0)
```

### Training One Autoencoder at a Time

Rather than training the whole stacked autoencoder in one go like we just did, it is possible to train one shallow autoencoder at a time, then stack all of them into a single stacked autoencoder. **This technique is not used much these days**

## Convolutional Autoencoders

If you are dealing with images then the autoencoders we have seen so far will not work well (unless the images are very small). Convolutional neural networks are far better suited than dense networks to work with images. So if you want to build an autoencoder for images (e.g. for unsupervised pretraining or dimensionality reduction), you will need to build a *convolutional encoder*.  The encoder is a regular CNN composed of convoluational layers and pooling layers. It typically reduces the spatial dimensionality of the inputs (i.e. height and width) while increasing the depth (i.e. the number of feature maps). The decoder must do the reverse (upscale the image and reduce its depth back to the originall dimensions), and for this you can transpose convolutional layers (alternatively, you could combine upsampling layers with convolutional layers). Below is a simple convolutional autoencoder for Fashion MNIST

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers


# -----------------------------
# 1. Load and prepare Fashion MNIST
# -----------------------------

(X_train_full, y_train_full), (X_test, y_test) = keras.datasets.fashion_mnist.load_data()

X_train_full = X_train_full.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Add channel dimension: (28, 28) -> (28, 28, 1)
X_train_full = X_train_full[..., tf.newaxis]
X_test = X_test[..., tf.newaxis]

X_train = X_train_full[:-5000]
X_val = X_train_full[-5000:]

y_train = y_train_full[:-5000]
y_val = y_train_full[-5000:]


# -----------------------------
# 2. Custom rounded accuracy metric
# -----------------------------

class RoundedAccuracy(tf.keras.metrics.Metric):
    def __init__(self, name="rounded_accuracy", **kwargs):
        super().__init__(name=name, **kwargs)
        self.binary_accuracy = tf.keras.metrics.BinaryAccuracy()

    def update_state(self, y_true, y_pred, sample_weight=None):
        y_true_rounded = tf.round(y_true)
        y_pred_rounded = tf.round(y_pred)

        self.binary_accuracy.update_state(
            y_true_rounded,
            y_pred_rounded,
            sample_weight=sample_weight
        )

    def result(self):
        return self.binary_accuracy.result()

    def reset_state(self):
        self.binary_accuracy.reset_state()


# -----------------------------
# 3. Build convolutional encoder
# -----------------------------

encoder = keras.Sequential([
    layers.Input(shape=(28, 28, 1)),

    layers.Conv2D(
        filters=16,
        kernel_size=3,
        padding="same",
        activation="selu",
        kernel_initializer="lecun_normal"
    ),
    layers.MaxPooling2D(pool_size=2),   # 28x28 -> 14x14

    layers.Conv2D(
        filters=32,
        kernel_size=3,
        padding="same",
        activation="selu",
        kernel_initializer="lecun_normal"
    ),
    layers.MaxPooling2D(pool_size=2),   # 14x14 -> 7x7

    layers.Conv2D(
        filters=64,
        kernel_size=3,
        padding="same",
        activation="selu",
        kernel_initializer="lecun_normal"
    )
], name="conv_encoder")


# -----------------------------
# 4. Build convolutional decoder
# -----------------------------

decoder = keras.Sequential([
    layers.Input(shape=(7, 7, 64)),

    layers.Conv2DTranspose(
        filters=32,
        kernel_size=3,
        strides=2,
        padding="same",
        activation="selu",
        kernel_initializer="lecun_normal"
    ),  # 7x7 -> 14x14

    layers.Conv2DTranspose(
        filters=16,
        kernel_size=3,
        strides=2,
        padding="same",
        activation="selu",
        kernel_initializer="lecun_normal"
    ),  # 14x14 -> 28x28

    layers.Conv2D(
        filters=1,
        kernel_size=3,
        padding="same",
        activation="sigmoid"
    )
], name="conv_decoder")


# -----------------------------
# 5. Build full autoencoder
# -----------------------------

conv_autoencoder = keras.Sequential([
    encoder,
    decoder
], name="conv_autoencoder")


# -----------------------------
# 6. Compile
# -----------------------------

conv_autoencoder.compile(
    optimizer=keras.optimizers.Nadam(),
    loss="binary_crossentropy",
    metrics=[RoundedAccuracy()]
)


# -----------------------------
# 7. Train
# -----------------------------

history = conv_autoencoder.fit(
    X_train,
    X_train,
    epochs=10,
    validation_data=(X_val, X_val)
)


# -----------------------------
# 8. Reconstruct images
# -----------------------------

X_reconstructions = conv_autoencoder.predict(X_test[:5], verbose=0)
```

```python
import matplotlib.pyplot as plt

def plot_reconstructions(model, data, starting_slice_index=0):
    """
    Plots 5 original images in row 1 and their reconstructions in row 2.

    Parameters
    ----------
    model : tf.keras.Model
        Trained autoencoder model.
    data : np.ndarray or tf.Tensor
        Image data shaped like (n_samples, 28, 28).
    starting_slice_index : int
        Starting index for selecting 5 images.
    """

    images = data[starting_slice_index : starting_slice_index + 5]
    reconstructions = model.predict(images, verbose=0)

    fig, axes = plt.subplots(2, 5, figsize=(10, 4))

    for i in range(5):
        # Original images
        axes[0, i].imshow(images[i].squeeze(), cmap="binary")
        axes[0, i].axis("off")
        axes[0, i].set_title(f"Original {starting_slice_index + i}")

        # Reconstructed images
        axes[1, i].imshow(reconstructions[i].squeeze(), cmap="binary")
        axes[1, i].axis("off")
        axes[1, i].set_title("Reconstructed")

    plt.tight_layout()
    plt.show()

plot_reconstructions(
    model=conv_autoencoder,
    data=X_val,
    starting_slice_index=0
)
```

## Recurrent Autoencoders

If you want to build an autoencoder for sequences, such as time series or text then recurrent neural networks may be better suited than dense networks. Building a *recurrent autoencoder* is a straightforward: the encoder is typically a sequence-to-vector RNN which compresses the input sequence down to a single vector. The decoder is a vector-to-sequence RNN that does the reverse. 

In the example below notice we use a RepeatVector layer as the first layer of the decoder, to ensure that its input vector gets fed to the decoder at each time step. 

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers


# -----------------------------
# Simple recurrent autoencoder
# Fashion MNIST: each image treated as a sequence
# 28 timesteps, 28 features per timestep
# -----------------------------

recurrent_encoder = keras.Sequential([
    layers.Input(shape=(28, 28)),
    layers.LSTM(100, activation="selu", return_sequences=False),
], name="recurrent_encoder")


recurrent_decoder = keras.Sequential([
    layers.Input(shape=(100,)),
    layers.RepeatVector(28),
    layers.LSTM(100, activation="selu", return_sequences=True),
    layers.TimeDistributed(
        layers.Dense(28, activation="sigmoid")
    )
], name="recurrent_decoder")


recurrent_autoencoder = keras.Sequential([
    recurrent_encoder,
    recurrent_decoder
], name="recurrent_autoencoder")


recurrent_autoencoder.compile(
    optimizer=keras.optimizers.Nadam(),
    loss="binary_crossentropy",
    metrics=[RoundedAccuracy()]
)


history = recurrent_autoencoder.fit(
    X_train,
    X_train,
    epochs=10,
    validation_data=(X_val, X_val)
)
```

Up to now, in order to force the autoencoder to learn interesting features, we have limited the size of the coding layer, making it undercomplete. There are actually many other kinds of constraints that can be used, including ones that allow the coding layer to be just as large as the inputs, or even larger, resulting in an *overcomplete autoencoder*.

## Denoising Autoencoders

Another way to force the autoencoder to learn useful features is to add nosie to its inputs, training it to recover the original, noise-free inputs. The noise can be pure Gaussian noise added to the inputs or it can be randomly switched off inputs, just like in dropout. The implementation is straightforward: it is a regular stacked autoencoder with an additional Dropout layer applied to the encoder's inputs (or you could use a GaussianNoise layer instead). Recall that the Dropout layer is only active during training (and so is the GaussianNoise layer). 

**Denoising autoencoders can be used for data visualization or unsupervised pretraining, like other autoencoders we've discussed so far, but they can also be used quite simply and efficiently to remove noise from images.**

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers


denoising_encoder = keras.Sequential([
    layers.Input(shape=(28, 28)),
    layers.Flatten(),

    # Optionally use Gaussian noise instead of dropout:
    # layers.GaussianNoise(0.2),

    layers.Dropout(0.2),

    layers.Dense(
        100,
        activation="selu",
        kernel_initializer="lecun_normal"
    ),
    layers.Dense(
        30,
        activation="selu",
        kernel_initializer="lecun_normal"
    )
], name="denoising_encoder")


denoising_decoder = keras.Sequential([
    layers.Input(shape=(30,)),

    layers.Dense(
        100,
        activation="selu",
        kernel_initializer="lecun_normal"
    ),
    layers.Dense(
        28 * 28,
        activation="sigmoid"
    ),
    layers.Reshape((28, 28))
], name="denoising_decoder")


denoising_autoencoder = keras.Sequential([
    denoising_encoder,
    denoising_decoder
], name="denoising_autoencoder")


denoising_autoencoder.compile(
    optimizer=keras.optimizers.Nadam(),
    loss="binary_crossentropy",
    metrics=[RoundedAccuracy()]
)

history = denoising_autoencoder.fit(
    X_train,
    X_train,
    epochs=10,
    validation_data=(X_val, X_val)
)
```

## Sparse Autoencoders

Another kind of constraint that often leads to good feature extraction is *sparsity*: by adding an appropriate term to the cost function, the autoencoder is pushed to reduce the number of active neurons in the coding layer. For example, it may be pushed to have on average 5% signficantly active neurons in the coding layer. This forces the autoencoder to represent each input as a combination of a small number of activations. As aresult, each neuron in the coding layer typically ends up representing a useful feature.

A simple approach is to use the sigmoid activation function in the coding layer (to constrain the codings to values between 0 and 1), use a large coding layer (e.g. with 300 units), and add some $l_{1}$ regularization to the coding layer's activations (the decoder is just a regular decoder).

The ActivityRegularization layer just returns its inputs, but as a side effect it adds a training loss equal to the sum of absolute values of its inputs (this layer only has an effect during training). Equivalently, you could remove the ActivityRegularization layer and set activity_regularizer=keras.regularizers.l1(1e-3) in the previous layer. This penalty will encourage the neural network to produce codings close to 0, but since it will also be penalized if it does not reconstruct the inputs correctly, it will have to output at least a few nonzero values. Using the $l_{1}$ norm rather than the $l_{2}$ norm will push the neural network to preserver the most important codings while eliminating the ones that are not needed for the input image.

**Another approach, which often yields better results, is to measure the actual sparsity of the coding layer at each training iteration, and penalize the model when the measured sparsity differs from the target sparsity. We do so by computing the average activation of each neuron in the coding layer, over the whole training batch. The batch size must not be too small, or else the mean will not be accurate.**

Once we have the mean activation per neuron, we want to penalize the neurons that are too active, or not active enough, by adding a *sparsity loss* to the cost function. For example, if we measure that a neuron has an average activation of 0.3, but the target sparsity is 0.1, it must be penalzied to activate less. A good cost function for this is Kullback-Leibler (KL) divergence which has much stronger gradients than the mean squared error.

In order to control the relative importance of the sparsity loss and the reconstruction loss, we can multiply the sparsity loss by a sparsity weight hyperparameter. If this weight is too high, the model will stick closely to the target sparsity, but it may not reconstruct the inputs properly, making the model useless. Conversely, if it is too low, the model will mostly ignore the sparsity objective and will not learn any interesting features.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers


activity_sparse_encoder = keras.Sequential([
    layers.Input(shape=(28, 28)),
    layers.Flatten(),

    layers.Dense(
        100,
        activation="selu",
        kernel_initializer="lecun_normal"
    ),

    layers.Dense(
        300,
        activation="sigmoid"
    ),

    layers.ActivityRegularization(l1=1e-3)
], name="activity_sparse_encoder")


activity_sparse_decoder = keras.Sequential([
    layers.Input(shape=(300,)),

    layers.Dense(
        100,
        activation="selu",
        kernel_initializer="lecun_normal"
    ),

    layers.Dense(
        28 * 28,
        activation="sigmoid"
    ),

    layers.Reshape((28, 28))
], name="activity_sparse_decoder")


activity_sparse_autoencoder = keras.Sequential([
    activity_sparse_encoder,
    activity_sparse_decoder
], name="activity_sparse_autoencoder")


activity_sparse_autoencoder.compile(
    optimizer=keras.optimizers.Nadam(),
    loss="binary_crossentropy",
    metrics=[RoundedAccuracy()]
)

history_activity_sparse = activity_sparse_autoencoder.fit(
    X_train,
    X_train,
    epochs=10,
    validation_data=(X_val, X_val)
)
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers


class KLDivergenceSparsityRegularizer(keras.regularizers.Regularizer):
    def __init__(self, target_sparsity=0.1, sparsity_weight=0.2):
        self.target_sparsity = target_sparsity
        self.sparsity_weight = sparsity_weight

    def __call__(self, activations):
        # Mean activation per coding neuron over the batch
        mean_activations = tf.reduce_mean(activations, axis=0)

        # Numerical stability
        eps = keras.backend.epsilon()
        mean_activations = tf.clip_by_value(mean_activations, eps, 1.0 - eps)

        target = self.target_sparsity

        kl_divergence = (
            target * tf.math.log(target / mean_activations)
            + (1.0 - target) * tf.math.log(
                (1.0 - target) / (1.0 - mean_activations)
            )
        )

        return self.sparsity_weight * tf.reduce_sum(kl_divergence)

    def get_config(self):
        return {
            "target_sparsity": self.target_sparsity,
            "sparsity_weight": self.sparsity_weight
        }


kl_sparse_encoder = keras.Sequential([
    layers.Input(shape=(28, 28)),
    layers.Flatten(),

    layers.Dense(
        100,
        activation="selu",
        kernel_initializer="lecun_normal"
    ),

    layers.Dense(
        300,
        activation="sigmoid",
        activity_regularizer=KLDivergenceSparsityRegularizer(
            target_sparsity=0.1,
            sparsity_weight=0.2
        )
    )
], name="kl_sparse_encoder")


kl_sparse_decoder = keras.Sequential([
    layers.Input(shape=(300,)),

    layers.Dense(
        100,
        activation="selu",
        kernel_initializer="lecun_normal"
    ),

    layers.Dense(
        28 * 28,
        activation="sigmoid"
    ),

    layers.Reshape((28, 28))
], name="kl_sparse_decoder")


kl_sparse_autoencoder = keras.Sequential([
    kl_sparse_encoder,
    kl_sparse_decoder
], name="kl_sparse_autoencoder")


kl_sparse_autoencoder.compile(
    optimizer=keras.optimizers.Nadam(),
    loss="binary_crossentropy",
    metrics=[RoundedAccuracy()]
)

history_kl_sparse = kl_sparse_autoencoder.fit(
    X_train,
    X_train,
    epochs=10,
    validation_data=(X_val, X_val),
    batch_size=256
)
```

## Variational Autoencoders

Another important category of autoencoders are *variational autoencoders*. They are quite different from all the autoencoders discussed so far. Particularly:

* They are *probabalistic autoencoders*, meaning that their outputs are partly determined by chance, even after training (as opposed to denoising autoencoders which use randomness only during training)
* Most importantly, they are *generative autoencoders*, meaning that they can generate new instances that look like they were sampled from teh training set.

As their name suggests, variational autoencoders perform varational Bayesian inference, which is an efficient way to perform approximate Bayesian inference.

Let's take a look at how they work. Instead of directly producing a coding for a given input, the encoder produces a *mean coding* $\mu$ and a standard deviation $\sigma$. The actual coding is then sampled randomly from a Gaussian distribution with mean $\mu$ and a standard deviation $\sigma$. After that the decoder decodes the sampled coding normally.

Although the inputs may have very a convoluted distribution, a variational autoencoder tends to produce codings that look as they they were sampled from a simple Gaussian distribtuion: during training the cost function pushes the codings to gradually migrate within the coding space (also called the *latent space*) to end up looking like a cloud of Gaussian points. One great consequence is that after training a variational autoencoder, you can very easily generate a new instance: just sample a random coding rom the Gaussian distribution, decode it, and voila.

Now let's look at the cost function. It is composed of two parts. The first part is the usual reconstruction loss that pushes the autoencoder to reproduce its inputs (we can use cross entropy for this). The second is the *latent loss* that pushes the autoencoder to have codings that look as though they were sampled from a simple Gaussian distribution: it is the KL divergence between the target distribution (i.e. the Gaussian distribution) and the actual distribution of the codings.

Let's start building a variational autoencoder for Fashion MNIST using the Functional API because the model is not entirely sequential. Note that the Dense layers that output codings_mean and codings_log_var have the same inputs. We then pass both codings_mean and codings_log_var to the Sampling layer. Finally, the variational_encoder model has three outputs, in case you want to inspect the values of codings_mean and codings_log_var. The only output we will use is the last one (codings). For the decoder we could have used the Sequential API instead of the Functional API, since it is really just a simple stack of layers. Finally, let's build the variational autoencoder model.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers


# -----------------------------
# 1. Sampling layer with KL loss
# -----------------------------

class Sampling(layers.Layer):
    def __init__(self, kl_weight=1 / 784, **kwargs):
        super().__init__(**kwargs)
        self.kl_weight = kl_weight

    def call(self, inputs):
        codings_mean, codings_log_var = inputs

        batch_size = tf.shape(codings_mean)[0]
        coding_size = tf.shape(codings_mean)[1]

        epsilon = tf.random.normal(shape=(batch_size, coding_size))

        codings = codings_mean + tf.exp(0.5 * codings_log_var) * epsilon

        kl_loss = -0.5 * tf.reduce_sum(
            1 + codings_log_var
            - tf.exp(codings_log_var)
            - tf.square(codings_mean),
            axis=-1
        )

        self.add_loss(self.kl_weight * tf.reduce_mean(kl_loss))

        return codings


# -----------------------------
# 2. Variational encoder
# -----------------------------

coding_size = 10

encoder_inputs = layers.Input(shape=(28, 28), name="encoder_input")

x = layers.Flatten()(encoder_inputs)

x = layers.Dense(
    150,
    activation="selu",
    kernel_initializer="lecun_normal"
)(x)

x = layers.Dense(
    100,
    activation="selu",
    kernel_initializer="lecun_normal"
)(x)

codings_mean = layers.Dense(coding_size, name="codings_mean")(x)
codings_log_var = layers.Dense(coding_size, name="codings_log_var")(x)

codings = Sampling(name="codings")([codings_mean, codings_log_var])

variational_encoder = keras.Model(
    inputs=encoder_inputs,
    outputs=[codings_mean, codings_log_var, codings],
    name="variational_encoder"
)


# -----------------------------
# 3. Variational decoder
# -----------------------------

decoder_inputs = layers.Input(shape=(coding_size,), name="decoder_input")

x = layers.Dense(
    100,
    activation="selu",
    kernel_initializer="lecun_normal"
)(decoder_inputs)

x = layers.Dense(
    150,
    activation="selu",
    kernel_initializer="lecun_normal"
)(x)

x = layers.Dense(
    28 * 28,
    activation="sigmoid"
)(x)

decoder_outputs = layers.Reshape((28, 28))(x)

variational_decoder = keras.Model(
    inputs=decoder_inputs,
    outputs=decoder_outputs,
    name="variational_decoder"
)


# -----------------------------
# 4. Full variational autoencoder
# -----------------------------

vae_inputs = layers.Input(shape=(28, 28), name="vae_input")

_, _, codings = variational_encoder(vae_inputs)
reconstructions = variational_decoder(codings)

variational_autoencoder = keras.Model(
    inputs=vae_inputs,
    outputs=reconstructions,
    name="variational_autoencoder"
)


# -----------------------------
# 5. Compile
# -----------------------------

variational_autoencoder.compile(
    optimizer=keras.optimizers.Nadam(),
    loss="binary_crossentropy",
    metrics=[RoundedAccuracy()]
)


# -----------------------------
# 6. Train
# -----------------------------

history_vae = variational_autoencoder.fit(
    X_train,
    X_train,
    epochs=25,
    batch_size=128,
    validation_data=(X_val, X_val)
)
```

### Generating Fashion MNIST Images

Now let's use this variational autoencoder to generate images that look like fashion items. All we need to do is sample random codings from a Gaussian distribution and decode them. Variational autoencoders make it possible to perform *semantic interpolation*: instead of interpolating two images at the pixel level (which would look as if the two images were overlaid), we can interpolate at the codings level. We first run both images through the encoder, then we interpolate the two codings we get, and finally we decode the interpolated codings to get the final image. 

For several years, variational autoencoders were quite popular, but GANs eventually took the lead, in particular because they are capable of generating much more realistic and crisp images. So let's turn our attention to GAN's

## Generative Adversarial Networks

Generative adversarial networks were proposed in 2014 and although the idea got researchers excited almost instantly it took a few years to overcome some of the difficulties of training GANs. It seems simple in hindsight: make neural networks compete against each other in the hope that this competition will push them to excel.

A GAN is composed of two neural networks:
* Generator
  > Takes a random distribution as input (typically Gaussian) and outputs some data (typically an image). You can think of the random inputs as the latent representation (i.e. codings) of the image to be generated. So, as you can see, the genrator offers the same functionality as a decoder in a variational autoencoder, and it can be used in the same way to generate new images (just feed it some Gaussian noise, and it outputs a brand-new image). However, it is trained very differently, as we will soon see
* Discriminator
  > Takes either a fake image from the generator or a real image from the training set as input and must guess whether the input image is fake or real.

During training, the generator and the discriminator have opposite goals: the discriminator tries to tell fake images from real images, while the generator tries to produce images that look real enough to trick the discriminator. Because the GAN is composed of two networks with different objectives, it cannot be trained like a regular neural network. Each training iteration is divided into two phases:
1. In the first phase, we train the discriminator. A batch of real images is sampled from the training set and is completed with an equal number of fake images produced by the generator. The labels are set to 0 for fake images and 1 for real images, and the discriminator is trained on this labeled batch for one step, using the binary cross-entropy loss. Importantly, backpropagation only optimizes the weights of the discriminator during this phase.
2. In the second phase, we train the generator. We first use it to produce another batch of fake images, and once again the discriminator is used to tell whether the images are fake or real. This time we do not add real images in the batch, and all the labels are set to 1 (real): in other words, we want the genrator to produce images that the discrimintaor will (wrongly) believe to be real. Crucially, the weights of the discriminator are frozen during this step, so backpropagation only affects teh weights of the generator. The generator never actually sees any real images, yet it gradually learns to produce convincing fake images. All it gets is the gradients flowing back through the discriminator. Fortunately, the better the discriminator gets the more information about the real images is contained in tehse secondhand gradients, so the genrator can make significant progress.

Let's go ahead and build a simple GAN for Fashion MNIST. The generator is similar to an autoencoder's decoder, and the discriminator is a regular binary classifier. For the second phase of each training iteration, we also need the full GAN model containing the generator followed by the discriminator. Next we will compile these models. As the discriminator is a binary classifier, we can naturally use the binary cross-entropy loss. The generator will only be trained through the gan model, so we do not need to compile it at all. The gan model is also a binary classifier, so it can use the binary cross-entropy loss. Importantly, the discriminator should not be trained during the second phase, so we make it non-trainable before compiling the gan model. The traininbg attribute is taken into account by Keras only when compiling a model, so after running this code the discriminator *is* trainable if we call its fit() method or its train_on_batch() method (which we will be using), while it is *not* trainable when we call these methods on the gan model.

Since the training loop is unusual, we cannot use the regular fit() method. Instead, we will write a cusom training loop. For this, we first need to create a Dataset to iterate through the images. As discussed earlier, you can see the two phases at each iteration:
1. In phase one we feed Gaussian noise to the generator to produce fake images, and we complete this batch by concatenating an equal number of real images. The targets y1 are set to 0 for fake images and 1 for real images. Then we train the discriminator on this batch. Note that we set the discriminator's trainable attribute to True: this is only to get rid of a warning that Keras displays when it notices that trainable is now False but was True when the model was compiled (or vice versa).
2. In phase two, we feed the GAN some Guassian noise. Its generator will start by producing fake images, then the discriminator will try to guess whether these images are fake or real, so the targets y2 are set to 1. Note that we set the trainable attribute to False, once again to avoid a warning.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt


# -----------------------------
# 1. Load and prepare Fashion MNIST
# -----------------------------

(X_train_full, y_train_full), (X_test, y_test) = keras.datasets.fashion_mnist.load_data()

X_train_full = X_train_full.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

X_train = X_train_full[:-5000]
X_val = X_train_full[-5000:]

# Add channel dimension: (28, 28) -> (28, 28, 1)
X_train = X_train[..., tf.newaxis]
X_val = X_val[..., tf.newaxis]
X_test = X_test[..., tf.newaxis]


# -----------------------------
# 2. Create TensorFlow dataset
# -----------------------------

batch_size = 32

train_dataset = (
    tf.data.Dataset.from_tensor_slices(X_train)
    .shuffle(buffer_size=10_000)
    .batch(batch_size, drop_remainder=True)
    .prefetch(tf.data.AUTOTUNE)
)


# -----------------------------
# 3. Build the generator
# -----------------------------

coding_size = 30

generator = keras.Sequential([
    layers.Input(shape=(coding_size,)),

    layers.Dense(7 * 7 * 128),
    layers.Reshape((7, 7, 128)),
    layers.BatchNormalization(),
    layers.Activation("selu"),

    layers.Conv2DTranspose(
        filters=64,
        kernel_size=5,
        strides=2,
        padding="same",
        activation="selu"
    ),  # 7x7 -> 14x14

    layers.Conv2DTranspose(
        filters=1,
        kernel_size=5,
        strides=2,
        padding="same",
        activation="sigmoid"
    )   # 14x14 -> 28x28
], name="generator")


# -----------------------------
# 4. Build the discriminator
# -----------------------------

discriminator = keras.Sequential([
    layers.Input(shape=(28, 28, 1)),

    layers.Conv2D(
        filters=64,
        kernel_size=5,
        strides=2,
        padding="same",
        activation=layers.LeakyReLU(negative_slope=0.2)
    ),

    layers.Dropout(0.4),

    layers.Conv2D(
        filters=128,
        kernel_size=5,
        strides=2,
        padding="same",
        activation=layers.LeakyReLU(negative_slope=0.2)
    ),

    layers.Dropout(0.4),

    layers.Flatten(),

    layers.Dense(1, activation="sigmoid")
], name="discriminator")


# -----------------------------
# 5. Compile discriminator and GAN
# -----------------------------

discriminator.compile(
    loss="binary_crossentropy",
    optimizer=keras.optimizers.Nadam(learning_rate=2e-4),
    metrics=["accuracy"]
)

discriminator.trainable = False

gan = keras.Sequential([
    generator,
    discriminator
], name="gan")

gan.compile(
    loss="binary_crossentropy",
    optimizer=keras.optimizers.Nadam(learning_rate=2e-4)
)


# -----------------------------
# 6. Custom GAN training loop
# -----------------------------
def train_gan(
    gan,
    dataset,
    batch_size,
    coding_size,
    n_epochs=10
):
    generator, discriminator = gan.layers

    for epoch in range(n_epochs):
        print(f"Epoch {epoch + 1}/{n_epochs}")

        for real_images in dataset:
            # -----------------------------
            # Phase 1: Train discriminator
            # -----------------------------

            noise = tf.random.normal(shape=(batch_size, coding_size))
            fake_images = generator(noise, training=True)

            X_fake_and_real = tf.concat([fake_images, real_images], axis=0)

            y_fake_and_real = tf.concat([
                tf.zeros((batch_size, 1)),
                tf.ones((batch_size, 1))
            ], axis=0)

            discriminator.trainable = True

            d_loss = discriminator.train_on_batch(
                X_fake_and_real,
                y_fake_and_real
            )

            # -----------------------------
            # Phase 2: Train generator
            # -----------------------------

            noise = tf.random.normal(shape=(batch_size, coding_size))

            y_generator = tf.ones((batch_size, 1))

            discriminator.trainable = False

            g_loss = gan.train_on_batch(
                noise,
                y_generator
            )

        print(f"  discriminator loss: {d_loss[0]:.4f}, discriminator acc: {d_loss[1]:.4f}, generator loss: {g_loss:.4f}")


# -----------------------------
# 7. Train
# -----------------------------
train_gan(
    gan=gan,
    dataset=train_dataset,
    batch_size=batch_size,
    coding_size=coding_size,
    n_epochs=30
)
```

```python
def plot_generated_images(generator, coding_size, n_images=10):
    noise = tf.random.normal(shape=(n_images, coding_size))
    generated_images = generator(noise, training=False)

    plt.figure(figsize=(10, 2))

    for i in range(n_images):
        plt.subplot(1, n_images, i + 1)
        plt.imshow(generated_images[i, :, :, 0], cmap="binary")
        plt.axis("off")

    plt.tight_layout()
    plt.show()


plot_generated_images(
    generator=generator,
    coding_size=coding_size,
    n_images=10
)
```

### The Difficulties of Training GANs

During training, the generator and the discriminator constantly try to outsmart each other, in a zero-sum game. As training advances the game may end up in a state that game theorists call a *Nash Equilibrium*, named after the mathematician John Nash. This is when no player would be better off changing their own strategy, assuming the other players do not change theirs (e.g. a predator chases its prey, the prey ties to escape, and neither would be better off changing their strategy). A GAN can only reach a single Nash equilibrium: that's when the generator produces perfectly realistic images, and the discriminator is forced to guess (50% real 50% fake). Nothing guarantees that the equilibrium will ever be reached, however.

The biggest difficulty is called *mode collapse*: this is when the generator's outputs gradually become less diverse. Suppose that the generator gets better at producing convincing shoes than any other class. It will fool the discriminator a bit more with shoes, and this will encourage it to produce even more images of shoes. Gradually, it will forget how to produce anything else. Meanwhile, the only fake images that the discriminator will see will be shoes, so it will also forget how to discrimintate fake images of other classes. Eventually, when the discriminator manages to discriminate the fake shoes from the real ones, the generator will be forced to move to another class. It may then become good at shirts, forgetting about shoes, and the discriminator will follow. The GAN may gradually cycle across a few classes, never really becoming very good at any of them.

Training may begin properly, then suddenly diverge for no apparent reason, due to tehse instabilities. And since many factors affect these complex dynamics, **GANs are very sensitive to the hyperparameters: you may have to spend a lot of effort fine-tuning them.** These problems have kept researchers very busy since 2014. A popular technique called *experience replay* consists in storing the images produced by the generator at each iteration in a replay buffer (gradually dropping older generated images) and training the discriminator using real images plus fkae images drawn from this buffer (rather than just fake images produced by the current generator). This reduces the chances that the discriminator will overfit the latest generator's outputs.

Another common technique is called *mini-batch discrimination*: it measures how similar images are across the batch and provides this statistic to the discriminator, so it can easily reject a whole batch of fake images that lack diversity. This encourages the generator to produce a greater variety of images, reducing the change of mode collapse. In short, this is still a very active field of research, and the dynamics of GANs are still not perfectly understood. But the good news is that great progress has been made so let's look at some of the most successful architectures, starting with deep convolutional GANs, which were the state of the art just a few years ago.

### Deep Convolutional GANs

Deep Convolutional GANs were popularized in 2015. In short:
* Replace any pooling layers with strided convolutions (in the discriminator) and transpose convolutions (in the generator)
* Use Batch Normalization in both the generator and the discrimintator, except in the generator's output layer and the discriminator's input layer
* Remove fully connected hidden layers for deeper architectures
* Use ReLU activiation in the generator for all layers except the outer layer, which should use tanh
* Use leaky ReLU activation in the discriminator for all layers.

In the example below, the generator takes codings of size 100, and it projects them to 6272 dimensions (7 * 7 * 128), and reshapes the result to get a 7 x 7 x 128 tensor. This tensor is batch normalized and fed to a transposed convolutioal layer with a stride of 2, which upsamples it from 7 x 7 to 14 x 14 and reduces its depth from 128 to 64. The result is batch normalized again and fed to another transposed convolutional layer with a stride of 2, which upsamples it from 14 x 14 to 28 x 28 and reduces the depth from 64 to 1. This layer uses the tanh activiation function, so the outputs will range from -1 to 1. For this reason, before training the GAN, we need to rescale the training set to that same range. We also need to reshape it to add the channel dimension. 

The discriminator looks much like a regular CNN for binary classification, except instead of using max pooling layers to downsample the image, we use strided convoluations (strides=2). Also note that we use the leaky ReLU activiation function. Overall, we respected the DCGAN guildlines, except we replaced the BatchNormalization layers in the discriminator with Dropout layers and we replaced ReLU with SELU in the generator.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt


# -----------------------------
# 1. Load and prepare data
# -----------------------------

(X_train_full, y_train_full), (X_test, y_test) = keras.datasets.fashion_mnist.load_data()

# Rescale to [-1, 1] because generator output uses tanh
X_train_full = X_train_full.astype("float32") / 127.5 - 1.0
X_test = X_test.astype("float32") / 127.5 - 1.0

# Train / validation split
X_train = X_train_full[:-5000]
X_val = X_train_full[-5000:]

# Add channel dimension: (28, 28) -> (28, 28, 1)
X_train = X_train[..., tf.newaxis]
X_val = X_val[..., tf.newaxis]
X_test = X_test[..., tf.newaxis]

batch_size = 32
coding_size = 100

train_dataset = (
    tf.data.Dataset.from_tensor_slices(X_train)
    .shuffle(10_000)
    .batch(batch_size, drop_remainder=True)
    .prefetch(tf.data.AUTOTUNE)
)


# -----------------------------
# 2. Build generator
# -----------------------------

generator = keras.Sequential([
    layers.Input(shape=(coding_size,)),

    layers.Dense(7 * 7 * 128),
    layers.Reshape((7, 7, 128)),
    layers.BatchNormalization(),
    layers.Activation("selu"),

    layers.Conv2DTranspose(
        64,
        kernel_size=5,
        strides=2,
        padding="same",
        use_bias=False
    ),
    layers.BatchNormalization(),
    layers.Activation("selu"),

    layers.Conv2DTranspose(
        1,
        kernel_size=5,
        strides=2,
        padding="same",
        activation="tanh"
    )
], name="generator")


# -----------------------------
# 3. Build discriminator
# -----------------------------

discriminator = keras.Sequential([
    layers.Input(shape=(28, 28, 1)),

    layers.Conv2D(
        64,
        kernel_size=5,
        strides=2,
        padding="same"
    ),
    layers.LeakyReLU(negative_slope=0.2),
    layers.Dropout(0.4),

    layers.Conv2D(
        128,
        kernel_size=5,
        strides=2,
        padding="same"
    ),
    layers.LeakyReLU(negative_slope=0.2),
    layers.Dropout(0.4),

    layers.Flatten(),
    layers.Dense(1, activation="sigmoid")
], name="discriminator")


# -----------------------------
# 4. Compile discriminator and GAN
# -----------------------------

discriminator.compile(
    loss="binary_crossentropy",
    optimizer=keras.optimizers.Adam(learning_rate=2e-4, beta_1=0.5),
    metrics=["accuracy"]
)

discriminator.trainable = False

gan = keras.Sequential([
    generator,
    discriminator
], name="dcgan")

gan.compile(
    loss="binary_crossentropy",
    optimizer=keras.optimizers.Adam(learning_rate=2e-4, beta_1=0.5)
)


# -----------------------------
# 5. Custom training loop
# -----------------------------

def train_dcgan(gan, dataset, batch_size, coding_size, n_epochs=30):
    generator, discriminator = gan.layers

    for epoch in range(n_epochs):
        print(f"Epoch {epoch + 1}/{n_epochs}")

        for real_images in dataset:
            # Phase 1: train discriminator
            noise = tf.random.normal(shape=(batch_size, coding_size))
            fake_images = generator(noise, training=True)

            X_fake_and_real = tf.concat([fake_images, real_images], axis=0)
            y_fake_and_real = tf.concat([
                tf.zeros((batch_size, 1)),
                tf.ones((batch_size, 1))
            ], axis=0)

            discriminator.trainable = True
            d_metrics = discriminator.train_on_batch(X_fake_and_real, y_fake_and_real)

            # Phase 2: train generator via GAN
            noise = tf.random.normal(shape=(batch_size, coding_size))
            y_generator = tf.ones((batch_size, 1))

            discriminator.trainable = False
            g_loss = gan.train_on_batch(noise, y_generator)

        d_loss = d_metrics[0]
        d_acc = d_metrics[1]
        g_loss_value = g_loss[0] if isinstance(g_loss, (list, tuple)) else g_loss

        print(
            f"  discriminator loss: {d_loss:.4f}, "
            f"discriminator acc: {d_acc:.4f}, "
            f"generator loss: {g_loss_value:.4f}"
        )


# -----------------------------
# 6. Train
# -----------------------------

train_dcgan(
    gan=gan,
    dataset=train_dataset,
    batch_size=batch_size,
    coding_size=coding_size,
    n_epochs=10
)


# -----------------------------
# 7. Plot generated images
# -----------------------------

def plot_generated_images(generator, coding_size, n_images=10):
    noise = tf.random.normal(shape=(n_images, coding_size))
    generated_images = generator(noise, training=False)

    # Rescale from [-1, 1] back to [0, 1]
    generated_images = (generated_images + 1.0) / 2.0

    plt.figure(figsize=(10, 2))
    for i in range(n_images):
        plt.subplot(1, n_images, i + 1)
        plt.imshow(generated_images[i, :, :, 0], cmap="binary")
        plt.axis("off")
    plt.tight_layout()
    plt.show()


plot_generated_images(generator, coding_size, n_images=10)
```

### Progressive Growing of GANs

DCGANs aren't perfect. When you try to generate very large images using DCGANs, you often end up with locally convincing features but overall inconsistencies such as shirts with one sleeve much longer than the other. To overcome this, an important technique was introduced in 2018 which generates small images at the beginning of training, then gradually adds convolutional layers to both the genrator and the discriminator to produce larger and larger images.

For example, when growing the generator's outputs from 4x4 to 8x8 an upsampling layer is added to the existing convolutional layer so that it outputs 8x8 feature maps, which are then fed to the new convolutional layer (which uses "same" padding and strides of 1 so it's outputs are also 8x8). This new layer is followed by a new output convolutional layer: this is a regular convolutional layer with kernel size 1 that projects the outputs down to the desired number of color channels (e.g. 3). To avoid breaking the trained weights of the first convolutional layer when the new convolutional layer is added, the final output is weighted sum of the original output layer (which now outpus 8x8 feature maps) and the new output layer. The weight of the new outputs is $\alpha$, while the weight of hte original outputs is $1 - \alpha$, and $\alpha$ is slowly increased from 0 to 1. In other words, the new convolutional layers are gradually faded in, while the original output layer is gradually faded out. A similar fade-in/fade-out technqiue is used when a new cconvolutional layer is added to the discriminator.

The 2018 paper also introduced several other techniques aimed at increasing the diversity of the outputs (to avoid mode collapse) and making training more stable:

1. Minibatch Standard Deviation Layer
   > Added near the end of the discriminator. For each position in the inputs, it computes the standard deviation across all channels and all instances in the batch. These standard deviations are then averaged across all points to get a single value. Finally, an extra freature map is added to each instance in the batch and filled with the computed value. How does this help? Well, if the generator produces images with little variety, then there will be a small standard deviation across freature maps in the discrimintaor. Thanks to this layer, the discriminator will have easy access to this statistic, making it less likely to be fooled by a generator that produces too little diversity. This will encourage the generator to produce more diverse outputs, reducing the risk of mode collapse.
2. Equalized Learning Rate
   > Initializes all weights using a simple Gaussian distribution with mean 0 and standard deviation 1 rather htan using He initialization. However, the weights are scaled down at runtime (i.e. every time the layer is executed) by the same factor as in He initialization: they are divided by $\sqrt{\frac{2}{n_{inputs}}}$ where $n_{inputs}$ is the number of inputs to the layer. **The paper demonstrated that this technique significantly improved the GAN's performance when using RMSProp, Adam, or other adaptive gradient optimizers**. By rescaling the weights as part of the model itself rather than just rescaling them upon initialization, this approach ensures that the dynamic range is the same for all paramters, throughout training, so they all learn at the same speed. This both speeds up and stabilizes training.
3. Pixelwise Normalization Layer
   > Added after each convolutional layer in the generator. It normalizes each activation based on all the activations in the same image and at the same location, but across all channels (dividing by the square root of the mean squared activation)

Evaluation is one of the big challenges when working with GANs: although it is possible to automatically evaluate the diversity of the generated images, juding their quality is a much trickier and subjective task. One technique is to use human raters, but this is costly and time-consuming. So the authors proposed to measure the similarity between the local image structure of the generated images and the training images, considering every scale. This idea led them to another groundbreaking innovation: StyleGANs.

### StyleGANs

The state of the art in high-resolution image generation was advanced once again by the same Nvidia team in a 2018 paper that popularized the StyleGAN architecture. **The authors used *style transfer* techniques in the generator to ensure that the generated images have the same local structure as the training images, at every scale, greatly improving the quality of the generated images. The discriminator and the loss function were not modified, only the generator.** It is composed of two networks:

1. Mapping Network
   > An eight-layer MLP that maps the latent representations **z** (i.e. the codings) to a vector **w**. This vector is then sent through mulitple *affine transformations** (i.e. Dense layers with no activation functions), which produces multiple vectors. These vectors control the style of the generated image at different levels, from fine-grained texture (e.g. hair color) to high-level features (e.g. adult or child). In short, the mapping network maps the codings to multiple style vectors.
2. Synthesis Network
   > Responsible for generating the images. It has a constant learned input (to be clear, this input will be constant after training, but during training it keeps getting tweaked by back propagation). It processes this input through multiple convolutional and upsampling layers, as earlier, but there are two twists: first, some noise is added to the input and to all the outputs of the convolutional layers (before the activation function). Second, each noise layer is followed by an *Adaptive Instance Normalization (AdaIN) layer: it standardizes each feature map independently (by subtracting the feature map's mean and dividing by its standard deviation), then it uses the style vector to determine the scale and offset of each feature map (the style vector contains one scale and one bias term for each feature map).

The idea of adding noise independently from the codings is very important. Some parts of an image are quite random, such as the exact position of each freckle or hair. In earlier GANs, this randomness had to either come from the codings or be some psuedorandom noise prodcued by the generator itself. If it came from the codings, it meant that the generator had to dedicate a significant portion of the codings' representational power to store noise, which is quite wasteful. Moreover, the noise had to be able to flow through the network and reach the final layers of the generator: this seems like an unnecessary constraint that probably slowed down training. And finally, some visual artifacts may appear because the same noise was used at different levels. If instead the generator tried to produce its own pseudorandom noise, this noise might not look very convincing, leading to more visual artifacts. Plus, part of the generator's weights would be dedicated to generating pseudorandom noise, which again seems wasteful. By adding extra noise inputs, all these issues are avoided; the GAN is able to use the provided noise to add the right amount of stochasticity to each part of the image.

The added noise is different for each level. Each noise input consists of a single feature map full of Gaussian noise, which is broadcast to all feature maps and scaled using learned per-feature scaling factors. Finally, StyleGAN uses a technique called *mixing regularization* (or *style mixing*), where a percentage of the generated images are produced using two different codings. This prevents the network from assuming that styles at adjacent levels are correlated, which in turn encourages locality in the GAN, mean that each style vector only affects a limited number of traits in the generated image. 

There is a such a wide variety of GANs out there that it would require a whole book to cover them all.

4. What are undercomplete and overcomplete autoencoders? What is the main risk of an excessively undercomplete autoencoder? What about the main risk of an overcomplete autoencoder?

**My Answer**

An undercomplete autoencoder has a latent/coding layer that is smaller than the input. An overcomplete autoencoder has a latent/coding layer that is equal to or larger than the input. The main risk of an excessively undercomplete autoencoder is information loss. The bottleneck may be too narrow, so the model cannot preserve enough meaningful variation to reconstruct the input well. It underfits. The main risk of an overcomplete autoencoder is learning the identity function: copying inputs to outputs without learning useful features. It can memorize or shortcut the task, producing low reconstruction loss but weak representations.

5. How do you tie weights in a stacked autoencoder? What is the point of doing so?

**My Answer**

You tie weights by making each decoder layer use the transpose of the corresponding encoder layer’s weight matrix, instead of learning its own independent weights. The point is to halve the number of learned weights, speed up training, reduce overfitting risk and force symmetry between encoding and decoding

7. What is a GAN? Can you name a few tasks where GANs can shine?

**My Answer**

Generative adversarial networks (GANs) use a generator and a discriminator, competing with one another during training, to generate synthetic instances (generator) and discriminate between real instances and synthetic (discriminator). Well trained GANs can generate believable synthetic data. One of the more cutting edge types of GANs is StyleGAN. It uses a mapping layer and a synthesis layer to generate different "style vectors", each of which is an embedding for a high or low level feature (like low level features like eyes or high level features like freckles) as well as inject noise into every convolutional layer. The output is a more refined generator that is capable of synthesizing very real outputs. GANs shine in image generation, super-resolution, inpainting, image-to-image translation, style transfer, and synthetic data augmentation. 8. What are the main difficulties when training GANs?

**My Answer**

The main difficulties when training GANs are instability, mode collapse, and training imbalance.

* Instability: the generator and discriminator can oscillate instead of converging.
* Mode collapse: the generator produces limited varieties of outputs.

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt
```

```python
# Load and split data
(X_train_full, y_train_full), (X_test, y_test) = keras.datasets.cifar10.load_data()

X_train_full = X_train_full.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

y_train_full = y_train_full.squeeze()
y_test = y_test.squeeze()

X_train = X_train_full
y_train = y_train_full
```

```python
# Build denoising autoencoder
coding_size = 256

# Encoder
encoder_inputs = layers.Input(shape=(32, 32, 3))

x = layers.GaussianNoise(0.1)(encoder_inputs)
# Alternative noise layer:
# x = layers.Dropout(0.2)(encoder_inputs)

x = layers.Conv2D(32, 3, padding="same", activation="relu")(x)
x = layers.MaxPooling2D(2, padding="same")(x)   # 32 -> 16

x = layers.Conv2D(64, 3, padding="same", activation="relu")(x)
x = layers.MaxPooling2D(2, padding="same")(x)   # 16 -> 8

x = layers.Conv2D(128, 3, padding="same", activation="relu")(x)
x = layers.Flatten()(x)

codings = layers.Dense(coding_size, activation="relu", name="coding_layer")(x)

encoder = keras.Model(
    inputs=encoder_inputs,
    outputs=codings,
    name="encoder"
)


# Decoder
decoder_inputs = layers.Input(shape=(coding_size,))

x = layers.Dense(8 * 8 * 128, activation="relu")(decoder_inputs)
x = layers.Reshape((8, 8, 128))(x)

x = layers.Conv2DTranspose(64, 3, strides=2, padding="same", activation="relu")(x)   # 8 -> 16
x = layers.Conv2DTranspose(32, 3, strides=2, padding="same", activation="relu")(x)   # 16 -> 32

decoder_outputs = layers.Conv2D(3, 3, padding="same", activation="sigmoid")(x)

decoder = keras.Model(
    inputs=decoder_inputs,
    outputs=decoder_outputs,
    name="decoder"
)


# Full autoencoder
autoencoder_inputs = layers.Input(shape=(32, 32, 3))
autoencoder_outputs = decoder(encoder(autoencoder_inputs))

denoising_autoencoder = keras.Model(
    inputs=autoencoder_inputs,
    outputs=autoencoder_outputs,
    name="denoising_autoencoder"
)

denoising_autoencoder.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3),
    loss="mse"
)

denoising_autoencoder.summary()
```

```python
# Train model
history_ae = denoising_autoencoder.fit(
    X_train,
    X_train,
    epochs=20,
    batch_size=128,
    validation_split=0.1
)
```

```python
# Visualize reconstructions
def plot_reconstructions(model, data, n_images=10, start=0):
    images = data[start:start + n_images]
    reconstructions = model(images, training=False)

    plt.figure(figsize=(n_images * 1.5, 3))

    for i in range(n_images):
        plt.subplot(2, n_images, i + 1)
        plt.imshow(images[i])
        plt.axis("off")

        plt.subplot(2, n_images, n_images + i + 1)
        plt.imshow(np.clip(reconstructions[i], 0, 1))
        plt.axis("off")

    plt.tight_layout()
    plt.show()


plot_reconstructions(
    model=denoising_autoencoder,
    data=X_test,
    n_images=10,
    start=0
)
```

10. Train a variational autoencoder on the image dataset of your choice, and use it to generate images. Alternatively, you can try to find an unlabeled dataset that you are interested in and see if you can generate new samples.

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

(X_train_full, y_train_full), (X_test, y_test) = keras.datasets.cifar10.load_data()

X_train_full = X_train_full.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

X_train = X_train_full

class Sampling(layers.Layer):
    def call(self, inputs):
        z_mean, z_log_var = inputs

        batch_size = tf.shape(z_mean)[0]
        latent_dim = tf.shape(z_mean)[1]

        epsilon = tf.random.normal(shape=(batch_size, latent_dim))

        return z_mean + tf.exp(0.5 * z_log_var) * epsilon
```

```python
latent_dim = 128

encoder_inputs = layers.Input(shape=(32, 32, 3))

x = layers.Conv2D(32, 3, strides=2, padding="same", activation="relu")(encoder_inputs)
x = layers.Conv2D(64, 3, strides=2, padding="same", activation="relu")(x)
x = layers.Conv2D(128, 3, strides=2, padding="same", activation="relu")(x)

x = layers.Flatten()(x)
x = layers.Dense(256, activation="relu")(x)

z_mean = layers.Dense(latent_dim, name="z_mean")(x)
z_log_var = layers.Dense(latent_dim, name="z_log_var")(x)

z = Sampling(name="z")([z_mean, z_log_var])

encoder = keras.Model(
    encoder_inputs,
    [z_mean, z_log_var, z],
    name="encoder"
)

encoder.summary()
```

```python
decoder_inputs = layers.Input(shape=(latent_dim,))

x = layers.Dense(4 * 4 * 128, activation="relu")(decoder_inputs)
x = layers.Reshape((4, 4, 128))(x)

x = layers.Conv2DTranspose(128, 3, strides=2, padding="same", activation="relu")(x)
x = layers.Conv2DTranspose(64, 3, strides=2, padding="same", activation="relu")(x)
x = layers.Conv2DTranspose(32, 3, strides=2, padding="same", activation="relu")(x)

decoder_outputs = layers.Conv2D(
    3,
    3,
    padding="same",
    activation="sigmoid"
)(x)

decoder = keras.Model(
    decoder_inputs,
    decoder_outputs,
    name="decoder"
)

decoder.summary()
```

```python
class VAE(keras.Model):
    def __init__(self, encoder, decoder, kl_weight=1e-4, **kwargs):
        super().__init__(**kwargs)
        self.encoder = encoder
        self.decoder = decoder
        self.kl_weight = kl_weight

        self.total_loss_tracker = keras.metrics.Mean(name="loss")
        self.reconstruction_loss_tracker = keras.metrics.Mean(name="reconstruction_loss")
        self.kl_loss_tracker = keras.metrics.Mean(name="kl_loss")

    @property
    def metrics(self):
        return [
            self.total_loss_tracker,
            self.reconstruction_loss_tracker,
            self.kl_loss_tracker
        ]

    def compute_losses(self, data, training=False):
        z_mean, z_log_var, z = self.encoder(data, training=training)
        reconstruction = self.decoder(z, training=training)

        reconstruction_loss = tf.reduce_mean(
            tf.reduce_sum(
                keras.losses.binary_crossentropy(data, reconstruction),
                axis=(1, 2)
            )
        )

        kl_loss = -0.5 * tf.reduce_mean(
            tf.reduce_sum(
                1 + z_log_var - tf.square(z_mean) - tf.exp(z_log_var),
                axis=1
            )
        )

        total_loss = reconstruction_loss + self.kl_weight * kl_loss

        return total_loss, reconstruction_loss, kl_loss

    def train_step(self, data):
        if isinstance(data, tuple):
            data = data[0]

        with tf.GradientTape() as tape:
            total_loss, reconstruction_loss, kl_loss = self.compute_losses(
                data,
                training=True
            )

        grads = tape.gradient(total_loss, self.trainable_weights)
        self.optimizer.apply_gradients(zip(grads, self.trainable_weights))

        self.total_loss_tracker.update_state(total_loss)
        self.reconstruction_loss_tracker.update_state(reconstruction_loss)
        self.kl_loss_tracker.update_state(kl_loss)

        return {
            "loss": self.total_loss_tracker.result(),
            "reconstruction_loss": self.reconstruction_loss_tracker.result(),
            "kl_loss": self.kl_loss_tracker.result()
        }

    def test_step(self, data):
        if isinstance(data, tuple):
            data = data[0]

        total_loss, reconstruction_loss, kl_loss = self.compute_losses(
            data,
            training=False
        )

        self.total_loss_tracker.update_state(total_loss)
        self.reconstruction_loss_tracker.update_state(reconstruction_loss)
        self.kl_loss_tracker.update_state(kl_loss)

        return {
            "loss": self.total_loss_tracker.result(),
            "reconstruction_loss": self.reconstruction_loss_tracker.result(),
            "kl_loss": self.kl_loss_tracker.result()
        }

    def call(self, inputs):
        _, _, z = self.encoder(inputs)
        return self.decoder(z)
```

```python
vae = VAE(
    encoder=encoder,
    decoder=decoder,
    kl_weight=1e-4,
    name="cifar10_vae"
)

vae.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3)
)

history = vae.fit(
    X_train,
    epochs=30,
    batch_size=128,
    validation_split=0.1
)
```

```python
def plot_vae_reconstructions(vae, data, n_images=10, start=0):
    images = data[start:start + n_images]
    reconstructions = vae(images, training=False)

    plt.figure(figsize=(n_images * 1.5, 3))

    for i in range(n_images):
        plt.subplot(2, n_images, i + 1)
        plt.imshow(images[i])
        plt.axis("off")

        plt.subplot(2, n_images, n_images + i + 1)
        plt.imshow(np.clip(reconstructions[i], 0, 1))
        plt.axis("off")

    plt.tight_layout()
    plt.show()


plot_vae_reconstructions(
    vae=vae,
    data=X_test,
    n_images=10,
    start=0
)
```

```python
def generate_images(decoder, latent_dim, n_images=16):
    random_latents = tf.random.normal(shape=(n_images, latent_dim))
    generated_images = decoder(random_latents, training=False)

    grid_size = int(np.sqrt(n_images))
    plt.figure(figsize=(grid_size * 2, grid_size * 2))

    for i in range(n_images):
        plt.subplot(grid_size, grid_size, i + 1)
        plt.imshow(np.clip(generated_images[i], 0, 1))
        plt.axis("off")

    plt.tight_layout()
    plt.show()


generate_images(
    decoder=decoder,
    latent_dim=latent_dim,
    n_images=16
)
```

11. Train a DCGAN to tackle the image dataset of your choice, and use it to generate images. Add experience replay and see if this helps. Turn it into a conditional GAN where you can control the generated class.

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt
from collections import deque


# =========================================================
# 1. Load and prepare CIFAR-10
# =========================================================

(X_train_full, y_train_full), (X_test, y_test) = keras.datasets.cifar10.load_data()

# Scale to [-1, 1] because generator uses tanh
X_train_full = X_train_full.astype("float32") / 127.5 - 1.0
X_test = X_test.astype("float32") / 127.5 - 1.0

y_train_full = y_train_full.astype("int32")
y_test = y_test.astype("int32")

batch_size = 128
latent_dim = 100
num_classes = 10

class_names = [
    "airplane", "automobile", "bird", "cat", "deer",
    "dog", "frog", "horse", "ship", "truck"
]

train_dataset = (
    tf.data.Dataset.from_tensor_slices((X_train_full, y_train_full))
    .shuffle(10000)
    .batch(batch_size, drop_remainder=True)
    .prefetch(tf.data.AUTOTUNE)
)


# =========================================================
# 2. Build conditional generator
# =========================================================

def build_generator(latent_dim=100, num_classes=10):
    noise_input = keras.Input(shape=(latent_dim,), name="noise_input")
    label_input = keras.Input(shape=(1,), dtype="int32", name="label_input")

    label_embedding = layers.Embedding(num_classes, latent_dim)(label_input)
    label_embedding = layers.Flatten()(label_embedding)

    x = layers.Concatenate()([noise_input, label_embedding])

    x = layers.Dense(4 * 4 * 256, use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)

    x = layers.Reshape((4, 4, 256))(x)

    x = layers.Conv2DTranspose(
        128, kernel_size=4, strides=2, padding="same", use_bias=False
    )(x)  # 4 -> 8
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)

    x = layers.Conv2DTranspose(
        64, kernel_size=4, strides=2, padding="same", use_bias=False
    )(x)  # 8 -> 16
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)

    x = layers.Conv2DTranspose(
        32, kernel_size=4, strides=2, padding="same", use_bias=False
    )(x)  # 16 -> 32
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)

    output_image = layers.Conv2D(
        3, kernel_size=3, padding="same", activation="tanh"
    )(x)

    return keras.Model([noise_input, label_input], output_image, name="generator")


# =========================================================
# 3. Build conditional discriminator
# =========================================================

def build_discriminator(num_classes=10):
    image_input = keras.Input(shape=(32, 32, 3), name="image_input")
    label_input = keras.Input(shape=(1,), dtype="int32", name="label_input")

    label_embedding = layers.Embedding(num_classes, 32 * 32)(label_input)
    label_embedding = layers.Flatten()(label_embedding)
    label_embedding = layers.Reshape((32, 32, 1))(label_embedding)

    x = layers.Concatenate(axis=-1)([image_input, label_embedding])

    x = layers.Conv2D(64, kernel_size=4, strides=2, padding="same")(x)
    x = layers.LeakyReLU(negative_slope=0.2)(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Conv2D(128, kernel_size=4, strides=2, padding="same")(x)
    x = layers.LeakyReLU(negative_slope=0.2)(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Conv2D(256, kernel_size=4, strides=2, padding="same")(x)
    x = layers.LeakyReLU(negative_slope=0.2)(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Flatten()(x)
    output = layers.Dense(1, activation="sigmoid")(x)

    return keras.Model([image_input, label_input], output, name="discriminator")


generator = build_generator(latent_dim=latent_dim, num_classes=num_classes)
discriminator = build_discriminator(num_classes=num_classes)


# =========================================================
# 4. Loss and optimizers
# =========================================================

bce = keras.losses.BinaryCrossentropy()

g_optimizer = keras.optimizers.Adam(learning_rate=2e-4, beta_1=0.5)
d_optimizer = keras.optimizers.Adam(learning_rate=2e-4, beta_1=0.5)


# =========================================================
# 5. Experience replay buffer
# =========================================================

replay_buffer = deque(maxlen=5000)

def add_to_replay(fake_images, fake_labels):
    fake_images_np = fake_images.numpy()
    fake_labels_np = fake_labels.numpy()

    for img, lab in zip(fake_images_np, fake_labels_np):
        replay_buffer.append((img.astype("float32"), lab.astype("int32")))

def sample_from_replay(n_samples):
    if len(replay_buffer) == 0 or n_samples <= 0:
        return None, None

    n_samples = min(n_samples, len(replay_buffer))
    idx = np.random.choice(len(replay_buffer), size=n_samples, replace=False)

    replay_images = np.stack([replay_buffer[i][0] for i in idx], axis=0)
    replay_labels = np.stack([replay_buffer[i][1] for i in idx], axis=0)

    return (
        tf.convert_to_tensor(replay_images, dtype=tf.float32),
        tf.convert_to_tensor(replay_labels, dtype=tf.int32)
    )


# =========================================================
# 6. One training step for discriminator
# =========================================================

@tf.function
def train_discriminator_step(real_images, real_labels, fake_images, fake_labels):
    real_targets = tf.ones((tf.shape(real_images)[0], 1)) * 0.9   # label smoothing
    fake_targets = tf.zeros((tf.shape(fake_images)[0], 1))

    with tf.GradientTape() as tape:
        real_preds = discriminator([real_images, real_labels], training=True)
        fake_preds = discriminator([fake_images, fake_labels], training=True)

        d_loss_real = bce(real_targets, real_preds)
        d_loss_fake = bce(fake_targets, fake_preds)
        d_loss = d_loss_real + d_loss_fake

    grads = tape.gradient(d_loss, discriminator.trainable_weights)
    d_optimizer.apply_gradients(zip(grads, discriminator.trainable_weights))

    real_acc = tf.reduce_mean(tf.cast(real_preds >= 0.5, tf.float32))
    fake_acc = tf.reduce_mean(tf.cast(fake_preds < 0.5, tf.float32))

    return d_loss, real_acc, fake_acc


# =========================================================
# 7. One training step for generator
# =========================================================

@tf.function
def train_generator_step(batch_size, latent_dim, num_classes):
    noise = tf.random.normal((batch_size, latent_dim))
    sampled_labels = tf.random.uniform(
        (batch_size, 1), minval=0, maxval=num_classes, dtype=tf.int32
    )

    misleading_targets = tf.ones((batch_size, 1))

    with tf.GradientTape() as tape:
        fake_images = generator([noise, sampled_labels], training=True)
        preds = discriminator([fake_images, sampled_labels], training=False)
        g_loss = bce(misleading_targets, preds)

    grads = tape.gradient(g_loss, generator.trainable_weights)
    g_optimizer.apply_gradients(zip(grads, generator.trainable_weights))

    return g_loss


# =========================================================
# 8. Training loop
# =========================================================

def train_conditional_dcgan(
    dataset,
    epochs=30,
    latent_dim=100,
    num_classes=10,
    use_replay=True,
    replay_fraction=0.5
):
    for epoch in range(epochs):
        d_losses = []
        g_losses = []
        real_accs = []
        fake_accs = []

        for real_images, real_labels in dataset:
            real_labels = tf.cast(real_labels, tf.int32)
            cur_batch_size = real_images.shape[0]

            if use_replay and len(replay_buffer) > 0:
                n_replay = int(cur_batch_size * replay_fraction)
            else:
                n_replay = 0

            n_current_fake = cur_batch_size - n_replay

            # Current fake images
            noise = tf.random.normal((n_current_fake, latent_dim))
            fake_labels_current = tf.random.uniform(
                (n_current_fake, 1),
                minval=0,
                maxval=num_classes,
                dtype=tf.int32
            )
            fake_images_current = generator(
                [noise, fake_labels_current],
                training=False
            )

            # Save current fake images to replay buffer
            add_to_replay(fake_images_current, fake_labels_current)

            # Replay fake images
            replay_images, replay_labels = sample_from_replay(n_replay)

            if replay_images is not None:
                fake_images_for_d = tf.concat(
                    [fake_images_current, replay_images], axis=0
                )
                fake_labels_for_d = tf.concat(
                    [fake_labels_current, replay_labels], axis=0
                )
            else:
                fake_images_for_d = fake_images_current
                fake_labels_for_d = fake_labels_current

            # Train discriminator
            d_loss, real_acc, fake_acc = train_discriminator_step(
                real_images,
                real_labels,
                fake_images_for_d,
                fake_labels_for_d
            )

            # Train generator
            g_loss = train_generator_step(
                cur_batch_size,
                latent_dim,
                num_classes
            )

            d_losses.append(float(d_loss))
            g_losses.append(float(g_loss))
            real_accs.append(float(real_acc))
            fake_accs.append(float(fake_acc))

        print(
            f"Epoch {epoch + 1}/{epochs} | "
            f"d_loss={np.mean(d_losses):.4f} | "
            f"g_loss={np.mean(g_losses):.4f} | "
            f"real_acc={np.mean(real_accs):.4f} | "
            f"fake_acc={np.mean(fake_accs):.4f} | "
            f"replay_size={len(replay_buffer)}"
        )


# =========================================================
# 9. Train
# =========================================================

train_conditional_dcgan(
    dataset=train_dataset,
    epochs=30,
    latent_dim=latent_dim,
    num_classes=num_classes,
    use_replay=True,      # set False to compare without replay
    replay_fraction=0.5
)


# =========================================================
# 10. Generate images for a chosen class
# =========================================================

def generate_class_images(generator, class_id, n_images=10, latent_dim=100):
    noise = tf.random.normal((n_images, latent_dim))
    labels = tf.fill((n_images, 1), tf.cast(class_id, tf.int32))

    generated = generator([noise, labels], training=False)

    # back to [0, 1] for plotting
    generated = (generated + 1.0) / 2.0
    generated = tf.clip_by_value(generated, 0.0, 1.0)

    plt.figure(figsize=(n_images * 1.5, 2))
    for i in range(n_images):
        plt.subplot(1, n_images, i + 1)
        plt.imshow(generated[i])
        plt.axis("off")
    plt.suptitle(f"Generated class: {class_names[class_id]}")
    plt.tight_layout()
    plt.show()


# Example: generate dogs
generate_class_images(generator, class_id=5, n_images=10, latent_dim=latent_dim)


# =========================================================
# 11. Generate one row per class
# =========================================================

def plot_one_per_class(generator, latent_dim=100):
    plt.figure(figsize=(12, 12))

    for class_id in range(num_classes):
        noise = tf.random.normal((1, latent_dim))
        label = tf.constant([[class_id]], dtype=tf.int32)

        image = generator([noise, label], training=False)
        image = (image + 1.0) / 2.0
        image = tf.clip_by_value(image, 0.0, 1.0)

        plt.subplot(2, 5, class_id + 1)
        plt.imshow(image[0])
        plt.title(class_names[class_id])
        plt.axis("off")

    plt.tight_layout()
    plt.show()


plot_one_per_class(generator, latent_dim=latent_dim)
```
